## Question 1: Multiplication Table

- Asks the user for a number.
- Uses a for loop to print the multiplication table of that number up to 10.

```python
# Ask the user for a number
num = int(input("Enter a number: "))

# Print the multiplication table up to 10 using a for loop
print(f"\nMultiplication table of {num}:")
for i in range(1, 11):
    print(f"{num} x {i} = {num * i}")
```

## Question 2: Simple Calculator Using Functions

- Defines separate functions for add, subtract, multiply, and divide.
- Asks the user to input two numbers and choose an operation.
- Calls the correct function and prints the result.

```python
# Define separate functions for each operation
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Error: Division by zero is not allowed."
    return a / b

# Ask the user for two numbers
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

# Ask the user to choose an operation
print("\nChoose an operation:")
print("1. Add")
print("2. Subtract")
print("3. Multiply")
print("4. Divide")
choice = input("Enter your choice (1/2/3/4): ")

# Call the correct function based on the user's choice
if choice == "1":
    result = add(num1, num2)
    print(f"\nResult: {num1} + {num2} = {result}")
elif choice == "2":
    result = subtract(num1, num2)
    print(f"\nResult: {num1} - {num2} = {result}")
elif choice == "3":
    result = multiply(num1, num2)
    print(f"\nResult: {num1} * {num2} = {result}")
elif choice == "4":
    result = divide(num1, num2)
    print(f"\nResult: {num1} / {num2} = {result}")
else:
    print("Invalid choice. Please select 1, 2, 3, or 4.")
```
